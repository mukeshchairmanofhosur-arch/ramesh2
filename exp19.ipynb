```python
import pandas as pd
import seaborn as sns

df = pd.read_csv("Book1.csv")

df.head()
```

```python
X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = df[['species']]
```

```python
y=df[['species']]
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_data = pd.DataFrame(scaler.fit_transform(X),columns=X.columns)
print(scaled_data.head())
```

```python
sns.heatmap(scaled_data.corr(), cmap='viridis', annot=True)
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components=4)
pca.fit(scaled_data)
X_pca = pca.transform(scaled_data)
X_pca = pd.DataFrame( X_pca,columns=['PC1', 'PC2', 'PC3', 'PC4'])
print(X_pca.head())

sns.heatmap(X_pca.corr(), cmap='viridis', annot=True)
```
